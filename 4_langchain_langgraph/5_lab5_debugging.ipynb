```python
from langchain_deepseek import ChatDeepSeek
from pydantic import BaseModel, Field

class EvaluatorOutput(BaseModel):
    feedback: str = Field(description="Feedback on the assistant's response")
    success_criteria_met: bool = Field(description="Whether the success criteria have been met")
    user_input_needed: bool = Field(
        description="True if the assistant has a question, needs clarification, or is stuck and needs the user"
    )

evaluator_model = ChatDeepSeek(
    model="deepseek-v4-flash",
    extra_body={"thinking": {"type": "disabled"}},
)

evaluator = evaluator_model.with_structured_output(EvaluatorOutput)

result = evaluator.invoke("Evaluate this: the task was completed successfully and criteria were met.")
print(result)
print(type(result))  # should be an EvaluatorOutput instance, not a raw string/dict
```
